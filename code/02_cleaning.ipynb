# 02_cleaning.ipynb

This notebook follows the in-class data cleaning pipeline:

1. Inspect  
2. Diagnose  
3. Missing values  
4. Duplicates  
5. Type conversion  
6. Validate and export

**Input:** `问卷数据.csv`  
**Output:** `questionnaire_cleaned.csv`


```python

import pandas as pd
import numpy as np

raw_path = r"/mnt/data/问卷数据.csv"
df_raw = pd.read_csv(raw_path, encoding="gb2312")
print("Loaded raw data:", df_raw.shape)
df_raw.head()
```

## 1. Inspect

```python
print("Shape:", df_raw.shape)
display(df_raw.head())
print("\nInfo:")
df_raw.info()
print("\nDtypes:")
display(df_raw.dtypes)
```

## 2. Diagnose

```python
print("Missing values by column:")
display(df_raw.isnull().sum().to_frame("missing_count"))

print("\nDuplicate rows:", df_raw.duplicated().sum())

for col in df_raw.columns:
    print(f"\nColumn: {col}")
    print(df_raw[col].dropna().astype(str).head(5).tolist())
```

### Diagnosis notes
- The file contains three empty columns named `Unnamed:*`.
- The first row is not a real response. It repeats question identifiers and should be removed.
- Numeric survey answers are stored as text and need conversion.
- The experimental group column can be used to derive `design_source` and `added_value`.


## 3–5. Clean: missing values, duplicates, type conversion

```python
df = df_raw.copy()

# Drop empty columns
df = df.loc[:, ~df.columns.str.contains(r"^Unnamed")]

# Remove the non-response header/code row
df = df[df["随机元素"] != "随机元素"].copy()

# Rename columns to analysis-friendly names
rename_map = {
    "随机元素": "group",
    "您是否同意参加本次研究？": "consent",
    "您线上购买服饰的频率？": "shopping_freq",
    "您是否购买过与 AI 相关的产品或服务？": "ai_purchase",
    "您的年龄": "age_group",
    "如果由您决定，您最多愿意为这件 T-shirt 支付多少钱？（0-500）": "wtp_open",
    "我愿意为这件 T-shirt 支付较高的价格。": "wtp_1",
    "我认为这件 T-shirt 值得这个价格。": "wtp_2",
    "我认为这种设计来源是可靠的。": "trust_1",
    "我相信这种设计来源能够完成高质量的产品设计。": "trust_2",
    "总体而言，我信任这种设计来源": "trust_3",
    "我对人工智能（AI）有基本了解": "fam_1",
    "我平时经常接触 AI 工具或应用": "fam_2",
    "我比较熟悉 AI 在产品设计中的应用": "fam_3",
    "我认为这件 T-shirt 是由 AI 完成设计的。": "mc_ai",
    "我认为这件 T-shirt 属于附加价值较高的产品。": "mc_value",
}
df = df.rename(columns=rename_map)

# Convert numeric columns
numeric_cols = [
    "wtp_open", "wtp_1", "wtp_2",
    "trust_1", "trust_2", "trust_3",
    "fam_1", "fam_2", "fam_3",
    "mc_ai", "mc_value"
]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Create experimental condition variables
df["design_source"] = df["group"].map({
    "实验情境1": 1,
    "实验情境2": 1,
    "实验情境3": 0,
    "实验情境4": 0
})
df["added_value"] = df["group"].map({
    "实验情境1": 1,
    "实验情境2": 0,
    "实验情境3": 1,
    "实验情境4": 0
})

# Convert some columns to ordered categories
df["age_group"] = pd.Categorical(
    df["age_group"],
    categories=["18-22", "23-27", "28-32", "33及以上"],
    ordered=True
)
df["shopping_freq"] = pd.Categorical(
    df["shopping_freq"],
    categories=["很少", "偶尔", "有时", "经常", "非常频繁"],
    ordered=True
)
df["ai_purchase"] = pd.Categorical(
    df["ai_purchase"],
    categories=["否", "不确定", "是"],
    ordered=True
)

# Construct scale scores
df["wtp_score"] = df[["wtp_1", "wtp_2"]].mean(axis=1)
df["trust_score"] = df[["trust_1", "trust_2", "trust_3"]].mean(axis=1)
df["ai_familiarity"] = df[["fam_1", "fam_2", "fam_3"]].mean(axis=1)

# Reset index after cleaning
df = df.reset_index(drop=True)

print("Cleaned shape:", df.shape)
display(df.head())
```

## 6. Validate

```python
print("Final shape:", df.shape)
print("\nMissing values:")
display(df.isnull().sum().to_frame("missing_count"))

print("\nDuplicate rows:", df.duplicated().sum())

print("\nDtypes:")
display(df.dtypes)

df.to_csv(r"/mnt/data/questionnaire_cleaned.csv", index=False, encoding="utf-8-sig")
print("Saved cleaned file to: /mnt/data/questionnaire_cleaned.csv")
```

## Cleaning summary
- Removed 1 non-response header row
- Dropped 3 fully empty `Unnamed:*` columns
- Converted rating variables and open-ended WTP to numeric
- Created:
  - `design_source` (1 = AI, 0 = professional design team)
  - `added_value` (1 = high, 0 = low)
  - `wtp_score`
  - `trust_score`
  - `ai_familiarity`
- Exported the cleaned dataset for downstream analysis
